```python
#agglomerative clustering

# Import necessary libraries
import numpy as np
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Step 1: Create some sample data
X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [8, 7],
    [9, 8],
    [10, 9],
    [11, 11],
    [20,80],
    [25, 80]
])

# Step 2: Create and fit the Agglomerative Clustering model
# n_clusters → number of clusters to form
# linkage → method for merging ('ward', 'complete', 'average', 'single')
model = AgglomerativeClustering(n_clusters=3, linkage='single')
labels = model.fit_predict(X)

# Step 3: Print results
print("Cluster labels:", labels)

# Step 4: Visualize the clusters
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow')
# X[:, 0] → takes all the x-values (first column of X).
# X[:, 1] → takes all the y-values (second column of X).
plt.title("Agglomerative Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
